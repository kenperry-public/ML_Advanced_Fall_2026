## The headline: one change in slope

<img src="images/fig_rate_break.png" width=50%>

> **Acceleration** = the ratio of a capability index's growth rate now to its earlier rate. 2x means two years of progress in one.

- <span style="color:#2ca02c"><b>FACT</b></span> Anthropic's index (AECI, one score from 15 benchmarks of well-specified tasks) grew about **1.9x** faster since Oct 2025 (95% interval 1.5 to 2.8) than in Mar 2024 to Oct 2025; METR placed the change at 7 Oct 2025
- <span style="color:#2ca02c"><b>FACT</b></span> METR guesses about **80%** of the extra growth is due to AI


## Three numbers and a threshold, not to be confused

| Number | What it is | Who says |
|---|---|---|
| **1.9x** | how much faster the index grows | <span style="color:#2ca02c"><b>FACT</b></span> METR measured it |

> Anthropic's own policy has a separate **2x threshold**: a speed-up of overall AI R&D progress, *due to AI*, over three models in a row.
> <span style="color:#ff7f0e"><b>EXTRAPOLATION</b></span> **Arithmetic:** if 80% of the extra 0.9x is AI, the AI-only ratio is about 1 + 0.8 x 0.9 = 1.7x (crude: the 80% is an Oct 2026 figure).

## What if the rate kept doubling? Repeated halving of the doubling time

<img src="images/fig_halving.png" width=50%>

- <span style="color:#2ca02c"><b>FACT</b></span> METR sees no clear evidence of repeated acceleration
- <span style="color:#ff7f0e"><b>EXTRAPOLATION</b></span> Curve C is a limit case, not a claim we found anyone making
- <span style="color:#ff7f0e"><b>EXTRAPOLATION</b></span> **Toy-model assumption (ours):** capability is 2^height, so height = doublings so far and slope = doublings per year; the real index is a benchmark score, not this
- <span style="color:#ff7f0e"><b>EXTRAPOLATION</b></span> **A:** one doubling per 2 years. **B:** the rate doubles once (one per year). **C:** the rate doubles after *each* doubling, so doubling times are 2, 1, 0.5, ... years
- $2 + 1 + \tfrac12 + \dots = 4$: infinitely many doublings in 4 years, so **C must break**


## What continuing would cost

- <span style="color:#ff7f0e"><b>EXTRAPOLATION</b></span> **E1:** at 2.4x a year: about <span class="tex2jax_ignore">&#36;</span>80B in 2032, <span class="tex2jax_ignore">&#36;</span>190B in 2033, <span class="tex2jax_ignore">&#36;</span>1 trillion in 2035
