## Name check

> **The term is weight-space programming.** It also circulates under a wrong name, "state-space programming".

1. **Weights**: the numbers inside a network that training sets (also called parameters)
2. **State-space model (SSM)**, such as Mamba: a *network design*; unrelated to this lecture
3. **Weight-space programming**: an *LLM sets the weights itself*, with no training run


## The idea in one picture

> **Instead of searching for good weights with gradient descent, a language model reasons about the weights and writes them.**

<img src="images/fig_training_vs_wsp.png" width=60%>

- **Why try it:** no training run, data set or loss is needed for a small task
- **The LLM** is GPT-6 Astra, the authors' frontier model, used as is
- **"Weights as text":** the numbers are pasted into the prompt


## Three modes: read, edit, write

<img src="images/fig_three_modes.png" width=64%>

- **Use 1 (this lecture): EDIT**, repairing a looping language model
- **Use 2 (this lecture): WRITE**, building a game-playing network from zeros
- READ is background: Astra named the game each trained network played, from weights alone


## Use 1 of 2: repairing a model stuck in loops

- **Problem:** Qwen3.5-0.8B, a small open model, often repeats itself until it runs out of tokens
    - 93% of its MATH-500 answers (competition math problems) hit the 32,768-token limit
- **Astra's tools:** it may run the model on practice problems
    - each run is a **forward pass**
    - a budget of 10, 25 or 50 passes, and a Python kernel over the weights
- **Diagnosis:** one *direction* (a repeated pattern in the model's internal numbers) in the MLP blocks of layers 15 and 19
- **Fix:** subtract that direction from the blocks' output weights

<img src="images/fig_edit_mechanism.png" width=52%>


## Use 1: what the repair bought

> **The repair worked only with 50 forward passes of feedback; with 10 or 25, the loops barely went away.**

<img src="images/fig_qwen_numbers.png" width=76%>

- Even the best repair left 33% of answers hitting the limit
- The MATH-500 jump is mostly **loops removed**, not new math skill
- Gemma 3 270M rarely loops (8.5%) and did not improve
    - authors: editing "could not add an ability the model lacked"


## Use 1: is this new?

> **It is mechanistic interpretability plus steering, done by an LLM and written into the weights.**

1. **Mechanistic interpretability:** find which internal parts cause a behavior
2. **Steering vector:** add or remove a direction in the model's internal numbers during a run
3. **Abliteration:** remove such a direction permanently (used to strip refusals)

- Critics: "basically a steering vector"
    - also close to ROME (2023), which edits stored facts in the weights
- Qwen's own model card already warns about thinking loops


## Use 2 of 2: the opposite direction, writing a network

> **Astra got the game's rules and source code, then wrote all the weights in one reply, starting from zeros.**

<img src="images/fig_breakout_program.png" width=74%>

- Breakout: a paddle-and-ball game; no data, loss, gradients or practice games
- The Breakout, Seaquest and Asterix networks were never run before scoring


## Use 2: how good were the written networks?

> **Breakout: about 20 times the best trained network. Asterix: about equal. Seaquest: about 60%.**

<img src="images/fig_scores.png" width=62%>

- Scores on a log scale
    - PPO is a standard reinforcement-learning trainer, shown at two training lengths
- Go (9x9, 1.48M weights): after 50 forward passes of practice games it beat a same-size PPO network trained on 20M games


## A program, or a discovered solution?

> **A written network is a program turned into weights: it imitates a human-style plan instead of discovering its own.**

| | Trained network | Written network |
|---|---|---|
| Who finds the solution | optimization, by trial | an LLM, by planning |
| Readable plan? | usually not | yes: landing point, then paddle |
| Can it surprise us? | yes, may find non-human tricks | not shown here |
| Needs | data, loss, compute | rules, source code, sometimes feedback |

- Tracr (2023) already compiles hand-written programs into transformer weights; the LLM automates it

## Three to be impressed by, three to be careful of

| Impressive | Careful | What critics say |
|---|---|---|
| No training compute or data | Tiny scale: edits to 0.8B, written nets up to 1.5M weights | Not new: ROME, hypernetworks, Tracr, steering |
| Surgical: 0.016% of weights changed | Needs rules or source, plus feedback passes | Weak baselines, e.g. a looping base model on MATH-500 |
| A plan precedes the weights, so a human can audit it | Cannot add abilities, only remove obstacles | One LLM, one unreviewed post; "self-improvement" overclaimed |


## What was reported, and what it sets up

- **Reported (one blog post; reproductions claimed on X, not checked here)**
    - an LLM can *edit and write* the weights of small networks
    - repairs target one localized fault; written networks follow a human-style plan
- **Open questions**
    - does it scale beyond toys and a 0.8B model?
    - what does an edit cost elsewhere (any hidden loss in other abilities)?
    - what does it mean for auditing and alignment?
