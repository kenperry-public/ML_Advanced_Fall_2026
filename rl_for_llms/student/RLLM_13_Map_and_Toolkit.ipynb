**Reading the map**

## Step 1 of 3: Placing a method. Others' samples (off-policy)

**Why this section?** A map helps only if you can place the next method you meet. Three questions place it:

1. whose samples; 2. who supplies the loss weights; 3. how fine.

| Method | What the weights name (what chose them) | Axis 2 | Axis 3 |
|---|---|---|---|
| SFT | the demonstrated tokens | reference | weight 1 on the dataset's one token |
| Classical distillation | the teacher's probabilities | reference | a distribution per token |
| DPO | $+c$ on the winner, $-c$ on the loser (a comparison chose them) | reward | one per completion (set by a pair) |

- DPO has a reference model (the anchor), but its weights come from a comparison: reward side
- Placements: our reading

## Step 1 of 3: Placing a method. Own samples (on-policy)

| Method | What the weights name (what chose them) | Axis 2 | Axis 3 |
|---|---|---|---|
| Self-distillation, RAFT | the kept tokens (a reward chose them) | across both: a **hybrid** (RLLM-9) | one per completion (0 or 1) |
| OPD, forward KL | the teacher's probabilities, at the student's prefixes | reference | a distribution per token |
| OPD, sampled token | a per-token signed number from the teacher, $u_{\tt}=\log\pi_{\text{T}}-\log\pi_\theta$ (RLLM-11) | reference (the teacher) | one per sampled token |
| GRPO, PPO | a signed reward for the completion | reward | GRPO: one per completion; PPO: one per sampled token |

- Placements: **our reading**

## Step 1 of 3: Two combinations RLLM-12 asked you to place

| Combination | Weights from | Place (our reading) |
|---|---|---|
| **Mixed loss**: the OPD loss plus an RL loss on a reward | the teacher (OPD term) and a reward (RL term) | own samples, across both columns; the RL term can reward outcomes the teacher never reaches |
| **Reward-filtered OPD**: OPD only on completions a reward keeps | the teacher | own samples, a hybrid like RLLM-9's; the weights stay the teacher's, so its loss has no term that pays for going above it (our reading) |

- Where the reward acts (the loss, or only the data choice) is what separates them

## Step 2 of 3: What do you hold? The Part 1 rule, completed

> **RLLM-6's rule covered Part 1's options. Part 2 adds two rows (demonstrations, a stronger model) and new options in the first rows, and says whose samples each row uses. A first pass (our reading), not a ranking.**

| You hold | Reach for | Whose samples |
|---|---|---|
| **a checker program** | RLVR (RL with verifiable rewards) with GRPO or PPO (RLLM-5, 6); or filter and imitate (RLLM-9) | own |
| **only people can judge** | a reward model and RLHF (RLLM-2); or DPO on pairs (RLLM-10) | own (stale); DPO's pairs: others' |
| **an AI judge with written principles** | RLAIF (RLLM-3), then as above | as above |
| **good demonstrations** | SFT (RLLM-8) | others' |
| **a stronger model** | classical distillation (RLLM-8), or OPD (RLLM-11) | others'; OPD: own |

## Step 2 of 3: One question for each thing you hold

- **Demonstrations** (SFT): are they **correct at every step**? Do many answers count as right (RLLM-1, RLLM-6)?
- **A checker** (a verifier): a reward on a whole completion, useful only if the model **already succeeds sometimes**
- **People's or an AI's judgments**: comparisons, for a reward model (RLHF) or for DPO
- **A stronger model**: is it **informative where the student struggles** (RLLM-12)?
- **Your generation budget**: does it cover fresh samples? The next slide
- Real pipelines combine them: **SFT first**, **RL after**. A reward is a **proxy**: the model learns what it rewards, loopholes included


## Step 2 of 3: What the first axis decides: fresh or stored samples

> **Own (fresh) samples cost generation and need a judge of what is good. Others' (stored) samples are cheap, and capped by the data.**

| Your situation | What axis 1 changes |
|---|---|
| no checker, no teacher, no judge | **others'** only: SFT. DPO on stored pairs needs a judge |
| a checker, and a generation budget | **own** is open: filter and imitate, GRPO, PPO |
| a tight generation budget | **others'**: stored demonstrations or pairs |
| a teacher, and you want the student's mistakes seen | **own** (OPD); classical distillation sees the teacher's prefixes only |

- **No evidence in this series that on-policy is better**: RLLM-7 gave each side's cost, RLLM-14 gives the checks


## Step 2 of 3: The decision table (1 of 3): SFT and the filter

**"Needs" is what you must hold; every row also needs prompts, and a held-out check of the result.**

| Method | Whose samples | Needs | Reach for it when | Watch out for |
|---|---|---|---|---|
| SFT | others' | demonstrations | you hold **good** ones | sees only **correct** prefixes |
| Filter, then imitate | own | a reward | you want a plain SFT step | **no** negative signal; the one-step identity needs a fixed bar; RAFT's rank rule: **not proved** |

- **Constant threshold**: keep a completion if its reward clears one fixed bar. **Weight 0**: the rest leave the batch (RLLM-9)

## Step 2 of 3: The decision table (2 of 3): rewards and pairs

| Method | Whose samples | Needs | Reach for it when | Watch out for |
|---|---|---|---|---|
| GRPO, PPO | own (stale) | a reward | you trust a reward and can afford sampling | **hackable** reward; sampling cost; whether **negative** weights beat filtering is **open** (RLLM-9) |
| RLHF, RLAIF | own (stale) | comparisons, or a judge | people or a model can say which is **better** | a learned **reward model** is hackable; one more model |
| DPO | others' (stored pairs); own (stale) online | preference pairs | you want no sampling loop | **fixed** pairs: stored states only |

## Step 2 of 3: The decision table (3 of 3): teachers, a reward added

| Method | Whose samples | Needs | Reach for it when | Watch out for |
|---|---|---|---|---|
| Classical distillation | others' | a teacher | its outputs are cheap to store | only the teacher's prefixes |
| OPD | own | a teacher | the teacher is **informative where the student struggles** (RLLM-12) | capped by the teacher; an open premise (RLLM-12) |
| OPD with a reward (a mixed loss, RLLM-12) | own | a teacher and a reward | one alone is **not enough** | more to **tune** |

- **Conditionals**: no method is free


## Step 3 of 3: What does each method's update equal? A ledger

**Why this section?** Two methods can look alike and still behave differently; this says how alike. Same update: one set of checks serves both. Same form only: different problems.

| Method | Looks like | Related to | How closely |
|---|---|---|---|
| Filter, constant threshold (RLLM-9) | SFT on kept completions | a policy gradient on a 0/1 reward | **updates**, one step, in expectation |
| RAFT, best of $G$ (RLLM-9) | SFT on the top completion per prompt | likely a policy gradient, rank-transformed reward; **not proved** | **form**, at most |
| DPO (RLLM-10) | a classification loss on pairs | the KL-regularized RL optimum, **if** Bradley-Terry holds (win chance depends on a score gap) | **solutions**, given enough data |

## Step 3 of 3: The ledger, continued

| Method | Looks like | Related to | How closely |
|---|---|---|---|
| OPD, forward KL (RLLM-11) | cross-entropy to a soft label | SFT, with the teacher's distribution | **form** |
| OPD, sampled token (RLLM-11) | a divergence to the teacher | a policy gradient with a per-token signed number from the teacher (our reading) | **updates** for the per-position loss (RLLM-11, our derivation); **form** only for the sequence objective |

**How closely**, strongest first:

- **Updates**: the same expected update in one step
- **Solutions**: the same best policy, given enough data
- **Form**: the same shape of loss, no claim about the optimum
- Always ask "equivalent to what, and in what sense?"
