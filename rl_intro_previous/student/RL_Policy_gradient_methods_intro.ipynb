# Policy-based methods/Policy gradient methods

**Why this session?** Value-based methods get the policy from values, and that route fails for continuous actions and very large action spaces. Here we learn the policy itself.

Value-based methods
- assign a value to a state (value function) OR action given a state (action value function)
- policy is *derived* from these values
    - choose the action leading to the greatest return
    - $\argmax{\act}$ to implement the policy

Policy-based methods
- by contrast, construct the policy $\pi$ directly
- as a parameterized (by parameters $\theta$) function

    $$
    \pi_\theta( \actseq | \stateseq ) = \prc{\actseq_\tt = \act}{\stateseq_\tt=\state}
    $$
    
i.e., the policy is a probability distribution of actions $\act$ , conditional on the state $\state$)

Policy-based methods are *necessary* in those cases in which Value-based methods are not possible:
- Continuous (versus discrete) action
    - the  $\argmax{\act}$ that implements policy in Value based methods is not possible
- Stochastic policy necessary
    - games against an adversary: when an adversary can take advantage of Agent predictability
- Aliased (partially observable) states: different situations that look the same to the agent
    - a stochastic policy can do better than any deterministic one
    
Policy-based methods are *desirable/preferable* when Value-based methods are impractical
- A large number of possible actions: the $\argmax{\act}$ over actions becomes expensive
- (large *state* spaces alone do not force a policy-based method: DQN handled them with a network)

| Scenario                               | Policy-Based Required | Policy-Based Desirable |
|:-------------------------------------- |:--------------------- |:---------------------- |
| Continuous action spaces               | Yes                   | Yes                    |
| Stochastic strategies needed           | Yes                   | Yes                    |
| Aliased or partially observable states | Yes                   | Yes                    |
| Large action spaces                    | Sometimes             | Yes                    |
| Discrete/simple environments           | No                    | Sometimes              |

Here is a brief comparison of Value-based and Policy-based methods.

| Aspect                     | Value-Based Methods                         | Policy-Based Methods                           |
|:-------------------------- |:------------------------------------------- |:---------------------------------------------- |
| Output                     | State/action value functions                | Directly parameterized policy                  |
| Policy Representation      | Implicit (via greedy/exploratory actions)   | Explicit (probability/distribution mapping)    |
| Learning Objective         | Value prediction loss minimization          | Expected return maximization (gradient ascent) |
| Typical Example Algorithms | DQN, Q-learning, SARSA                      | REINFORCE, PPO, vanilla policy gradient        |
| Action Space               | Discrete (practical)                        | Handles continuous and discrete                |
| Stochastic Policies        | Added by hand (epsilon-greedy)              | Natural/efficient                              |
| Exploration Strategies     | Decoupled from policy (e.g. epsilon-greedy) | Inherent (stochastic policy outputs)           |

## Policy Gradient methods

**Why this section?** To improve a policy by gradient ascent we must say what to ascend. The next slides build that objective and its gradient.

The predominant class of Policy based methods are those based on the *Policy Gradient* method.

Policy Gradient methods 
create a sequence of improving policies
$$
\pi_0, \ldots, \pi_p, \ldots
$$
by creating a sequence of improved parameter estimates
$$
\theta_0, \ldots, \theta_p, \ldots
$$
using Gradient Ascent on some objective function $J(\theta)$ to improve $\theta_p$
$$
\theta_{p+1} = \theta_p + \alpha * \nabla_\theta J(\theta_p)
$$
- gradient of an Objective Function $J(\theta)$
- with respect to parameters $\theta$



Since we are trying to maximize objective function $J(\theta)$ rather than minimize a loss objective
- we use Gradient Ascent rather than Gradient Descent
- hence we add the gradient rather than subtract it, in the update

[RL Book Chapt 13](http://incompleteideas.net/book/RLbook2020.pdf#page=343)

**Aside**

There are a few Policy based methods that *don't* use Policy Gradient
- in the module on Value based methods, we learned about Policy Iteration
- Policy iteration alternates
    - Policy Evaluation: improving the estimate of a Value function
    - Policy Improvement: improving the policy
        - use  $\argmax{\act}$ to implement the current policy 

Thus, Policy Iteration is both Value based and Policy based
- but does not evolve policy via gradients

## Stochastic policy and environment

**Why this section?** A gradient needs the policy to be a probability distribution. Here is where the randomness comes from, in the policy and in the environment.

With Value based methods
- the Environment can be stochastic
- but the policy it implements is usually deterministic (the greedy action; exploration is added by hand, e.g. $\epsilon$-greedy)
    -  $\argmax{\act}$ to implement the policy
    
With Policy Gradient methods
- the policy can be stochastic (action is a probability distribution)
    $$
    \pi( \act | \state; \theta ) = \pr{ \actseq_\tt = \act | \stateseq_\tt = \state, \theta_\tt = \theta }
    $$
- for example: Non-greedy policy that trades off Exploitation vs Exploration

The environment can *also* be stochastic
$$
\transp({ \state', \rew | \state, \act }) 
 = 
\transp({ \stateseq_{\tt+1}, \rewseq_{\tt+1} | \state = \stateseq_\tt, \act = \actseq_\tt })
$$


- the response $(\state', \rew)$ by the environment is not deterministic

A single observation of $(\state', \rew)$ is only a noisy sample of $\transp({ \state', \rew | \state, \act })$
- every model-free method must work from such samples
- the Policy Gradient Theorem will show that the gradient never needs $\transp$ itself

## Objective function

**Why this section?** Gradient ascent needs something to maximize: the expected return.

The performance measure $J(\theta)$ that we seek to maximize is the
- *expected value* (across each possible episode $\tau$) of 
- the return $\rlret_{0, \tau}$ from initial state $\stateseq_0$ of the episode

$$
J(\theta) = \Exp{\tau \sim \pi_\theta} ( \rlret_{0, \tau} )
$$

- $\tau \sim \pi_\theta$ means episodes generated by following the policy $\pi_\theta$
    - their probability $\prc{\tau}{\theta}$ is a function of the policy parameters $\theta$
- this is the value of the initial state, $\statevalfun_{\pi}(\stateseq_0)$, of the value-based sessions

Recall: the return from state $\stateseq_\tt$ of the episode is
$$
\begin{array}{lll} \\
\rlret_{\tt, \tau} & = & \sum_{k=0}^{|\tau|-\tt-1} {  \disc^k * \rewseq_{\tt+k+1} } \\
      & = & \rewseq_{\tt+1}  + \disc * \rlret_{\tt+1, \tau} \\
\end{array}
$$








## Taking the gradient of the Objective

**Why this section?** The objective is an expectation over episodes, and the environment's probabilities are unknown. How can we get its gradient at all?

For Gradient Ascent/Descent
- We need to be able to compute
$$ 
\nabla_\theta J(\theta_p)
$$

the gradient of the Objective w.r.t the parameters

However: there is an issue in computing the *gradient* of $J(\theta)$.

- Letting $
\pr{\tau ; \theta}
$
denote the probability of episode $\tau$ occurring
- the expectation can be re-written as a probability-weighted sum

$$
\Exp{\tau \sim \pr{\theta}} ( \rlret_{0, \tau} ) = \sum_\tau {  \pr{\tau ; \theta} * \rlret_{0, \tau} }
$$

We can compute trajectory probability $\prc{\tau}{\theta}$
- as a chained (multiplicative) probably of each step in the trajectory
$$
\prc{\tau}{\theta} = 
\pr{\stateseq_{\tau,0}}
*
\prod_{\tt=0}^{|\tau|-1} { 
\pi(\actseq_{\tau,\tt} | \stateseq_{\tau,\tt}) * \transp(\stateseq_{\tau, \tt+1}, \rewseq_{\tau,\tt+1} | \stateseq_{\tau,\tt}, \actseq_{\tau,\tt})
} 
$$

The issue is that $\pr{\tau ; \theta}$ depends on
- the Environment's response at each step of $\tau$
$$
\transp({ \stateseq_{\tt+1}, \rewseq_{\tt+1} | \state = \stateseq_\tt, \act = \actseq_\tt })
$$


which is *not available to us* under the assumption of *Model-free* methods

So, 
- how can we compute the gradient of an expectation
- when don't know the probability distribution


# Policy Gradient Theorem

**Why this section?** It answers the question just posed: the gradient can be written without the environment's probabilities. We state it first; the derivation follows as an aside.


The *Policy Gradient Theorem* tells us how to compute the Gradient of the Expectation.

Most importantly
- the Environment's Transition probability *does not* appear
- so this results in an operational way to compute the Gradient needed for maximization of $J(\theta)$

**Notes**

- The factor $\disc^\tt$ in the Return-to-go form (two slides on) is what makes it equal *in expectation* to the Episode Return form when $\disc < 1$; with $\disc = 1$ (no discounting) it disappears. Many references and implementations drop it, but then the gradient is a different (biased) estimate of the gradient of the discounted objective


We can write the Policy Gradient Theorem  in two equivalent forms (equal in expectation)

**Episode Return form**

$$
\nabla_\theta J(\theta) = 
\Exp{\tau \sim \pi_\theta} { 
\sum_{\tt=0}^{|\tau|-1} {
\nabla_\theta   \log \pi(\actseq_{\tau,\tt} | \stateseq_{\tau,\tt}) \, \,\rlret_{0,\tau}
}
} 
$$

where
$$\rlret_{0, \tau}$$
is the *episode return*: the return from the first step of episode $\tau$
 
This form is particularly useful
- when there is a *single reward* received at the end of the trajectory

**Notes**

- To clarify that states, actions, rewards, returns, etc. depend on the specific trajectory $\tau$
    - we add an extra subscript when necessary for clarification
$$
\rewseq_{\tau,\tt}
$$

**Return-to-go form**

$$
\nabla_\theta J(\theta) = \E_{\tau \sim \pi_\theta} \left[ \sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_{\tau,\tt} | \stateseq_{\tau,t}) \, \, \disc^\tt \, \rlret_{\tau,\tt} \right]
$$

where 
$$\rlret_{\tau,\tt}$$

is the return-to-go of the trajectory $\tau$ from step $\tt$ (state $\stateseq_{\tau, \tt}$), and $\disc^\tt$ is $1$ when $\disc = 1$.

This form is particularly useful
- when there are rewards at intermediate steps of the episode



Whichever way we write it
- the Policy Gradient Theorem is the foundation for all policy-based methods
- it tells us how to change the parameters $\theta$ of the Policy NN in a direction leading to optimality

We study the practical methods built on it in the next session.

**What one update does** (one state, a softmax policy over three actions)

- the parameters $\theta$ are the three *logits*, here $\theta = (0.4, 0, -0.4)$, and the second action is the one sampled; for a softmax, $\nabla_\theta \log \pi_\theta(\act) = \text{onehot}(\act) - \pi_\theta$
- the update is $\theta \leftarrow \theta + \alpha \, \rlret \, \nabla_\theta \log \pi_\theta(\act)$, with $\alpha = 0.5$ and $\rlret$ the return that followed the sampled action $\act$

<img src="images/policy_gradient_bars.png" width=95%>

- an action followed by a *positive* return becomes more likely, one followed by a *negative* return less likely; the other actions move the opposite way (the probabilities sum to $1$)
- if every return is positive, every sampled action is pushed up: the *expected* gradient is unchanged but the noise is large; a **baseline** fixes this (next session)


## The likelihood ratio trick

**Why this section?** The derivation needs one tool: turning the gradient of an expectation into an expectation of a gradient.

### Likelihood ratio trick

The likelihood ratio trick states that 
- for a parameterized probability distribution $p_\theta(x)$ 
- and a function $f(x)$:

the gradient of an expectation can be converted into an expectation of $f$ times the gradient of the log-probability (the *score function*).

It is a simple consequence of the Derivative of a Log rule of calculus.

$$
\begin{array}{llll} \\
\nabla_\theta \E_{x \sim p_\theta}[f(x)] & = & \nabla_\theta \int p_\theta(x) f(x) dx  & \text{convert expectation to integral}\\
& = &\int \nabla_\theta p_\theta(x) f(x) dx & \text{move grad inside the integral} \\
& = &\int f(x) \nabla_\theta p_\theta(x)  dx & \text{rearrange term} \\
& = &\int f(x) \left(p_\theta(x) \nabla_\theta \log p_\theta(x) \right)  dx & \text{log trick: } \\
& & & \nabla_\theta p_\theta(x) = p_\theta(x) \nabla_\theta \log p_\theta(x) \\
& = & \E_{x \sim p_\theta} \left[ f(x) \nabla_\theta \log p_\theta(x) \right] & \text{convert integral back to expectation} \\
\end{array}
$$

The "log trick" follows from the rules of calculus
$$
\begin{array}{llll} \\
\nabla_\theta \log p_\theta(x) & = & \frac{1}{p_\theta(x)} * \nabla_\theta p_\theta(x) & \text{Calculus: grad of log, chain rule} \\
\nabla_\theta  p_\theta(x) & = &  p_\theta(x) * \nabla_\theta  \log p_\theta(x) & \text{re-arranging terms} \\
\end{array}
$$

## Computing $\nabla_\theta J(\theta)$

**Why this section?** Putting the tool to work: the derivation of the theorem (for your own review).

Our first step will be to turn the expectation 
$$
\Exp{\tau \sim \prc{\tau}{\theta}} \,\, { \rlret_{0,\tau} }
$$
into a sum
$$
\sum_\tau \prc{\tau}{\theta} *  { \rlret_{0,\tau} }
$$

where
$$
\prc{\tau}{\theta}
$$
is the probability of trajectory $\tau$

Turning the expectation into a sum and taking the gradient of the expected Episode Return

$$
\begin{array}{llll} \\
\nabla_\theta
\sum_{\tau \sim \prc{\tau}{\theta}} \prc{\tau}{\theta} *  { \rlret_{0,\tau} } 
& = & \sum_{\tau \sim \prc{\tau}{\theta}}{ \nabla_\theta  \, \,\prc{\tau}{\theta} *  { \rlret_{0,\tau} }}  & \text{grad of a sum is sum of the grads} \\ 
& = & \sum_{\tau \sim \prc{\tau}{\theta}}{ (\prc{\tau}{\theta} \nabla_\theta  \log  \prc{\tau}{\theta}) \, \,\rlret_{0,\tau} } & \text{Likelihood ratio trick:} \\
& & & \nabla_\theta \prc{\tau}{\theta} = \prc{\tau}{\theta} * \nabla_\theta \log \prc{\tau}{\theta} \\
\end{array}
$$

We now substitute the chained probability previously derived for 
$$\prc{\tau}{\theta}$$
into the 
$$
\log \prc{\tau}{\theta}
$$

term above.



$$
\begin{array}{llll} \\
\log \prc{\tau}{\theta} & = & \log \left( 
\pr{\stateseq_{\tau,0}}
*
\prod_{\tt=0}^{|\tau|-1} { \pi(\actseq_{\tau,\tt} | \stateseq_{\tau,\tt}) * \transp(\stateseq_{\tau, \tt+1}, \rewseq_{\tau,\tt+1} | \stateseq_{\tau,\tt}, \actseq_{\tau,\tt})}  \right) & \text{definition of } \prc{\tau}{\theta} \\
& = & 
\log \pr{\stateseq_{\tau,0}}
+
  \sum_{\tt=0}^{|\tau|-1} {
  \left( \,
\log \pi(\actseq_{\tau,\tt} | \stateseq_{\tau,\tt}) + \log \transp(\stateseq_{\tau, \tt+1}, \rewseq_{\tau,\tt+1} | \stateseq_{\tau,\tt}, \actseq_{\tau,\tt})
\, \right)
} & \text{log of product is sum of logs}  \\
\\
\nabla_\theta \log \prc{\tau}{\theta} & = & 
\sum_{\tt=0}^{|\tau|-1} {
\nabla_\theta  \log \pi(\actseq_{\tau,\tt} | \stateseq_{\tau,\tt}) 
}  & \text{grad of sum is sum of grads} \\
& & &  \text{Transition Probability terms } \\
& & & \;\;\;\transp( \ldots) \\
& & & \text{are independent of } \theta \\
\end{array}
$$

The initial-state probability $\pr{\stateseq_{\tau,0}}$ does not depend on $\theta$ either, so it also drops out.


To summarize the proof:

$$
\begin{array}{llll} \\
\nabla_\theta \Exp{\tau \sim \pi_\theta} \,\, { \rlret_{0,\tau} } 
& = & \nabla_\theta \sum_\tau \prc{\tau}{\theta} *  { \rlret_{0,\tau} } & \text{turn the expectation into a sum weighted by episode probabilities} \\
& = & \sum_\tau{ \left(\prc{\tau}{\theta} \nabla_\theta  \log  \prc{\tau}{\theta} \right) \, \, *\rlret_{0,\tau} } & \text{Likelihood ratio trick} \\
& = & \sum_\tau{ }{ \left( \prc{\tau}{\theta} * 
    \sum_{\tt=0}^{|\tau|-1} {
\nabla_\theta  \log \pi(\actseq_{\tau,\tt} | \stateseq_{\tau,\tt}) } \right)
*  { \rlret_{0,\tau} }
} & \text{Transition Probability terms disappear from within the grad} \\
& & & \text{as shown above}\\
& = & \Exp{\tau \sim \prc{\tau}{\theta}}
{
\sum_{\tt=0}^{|\tau|-1} {
\nabla_\theta  \log \pi(\actseq_{\tau,\tt} | \stateseq_{\tau,\tt})  }
} \, \, *\rlret_{0,\tau}
& \text{turn back into an expectation} \\
& & &\text{involving each step of the episode} \\
\end{array}
$$

What about the expectation 
$\Exp{\tau \sim \prc{\tau}{\theta}}$ ?

The expectation can be approximated by *sampling* trajectories
- observe the *effect* of the Environment's distribution
- without *knowing* it

**Key result**

We can evaluate the Gradient *without knowing* the 
Environment Transition Probability terms $\transp(\ldots)$

**Notation alert**

The convention is to write the expectation 
- 
$
\Exp{\tau \sim \pi_\theta}
$
rather than
$
\Exp{\tau \sim \prc{\tau}{\theta}}
$

This is merely convention: they refer to the same distribution of episodes.

**In practice** we replace the expectation by an average over $N$ sampled episodes
$$
\nabla_\theta J(\theta) \approx \frac{1}{N} \sum_{i=1}^{N} \sum_{\tt=0}^{|\tau_i|-1} \nabla_\theta \log \pi_\theta(\actseq_{i,\tt} \mid \stateseq_{i,\tt}) \, \rlret_{0,\tau_i}
$$
and take a step $\theta \leftarrow \theta + \alpha \, \nabla_\theta J(\theta)$

This estimator is the core of **REINFORCE** (next session). It is *on-policy*: the episodes must come from the policy being improved.


### Why substituting the return-to-go $\rlret_{\tau,\tt}$ for the episode return $\rlret_{0,\tau}$ gives the same expectation

Consider the two equivalent forms for expressing the Policy Gradient Theorem

$$
\sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt) \cdot \rlret_{0,\tau}
$$

and

$$
\sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt) \cdot \disc^\tt \rlret_\tt
$$

where:
$$ \rlret_\tt = \sum_{k=\tt}^{|\tau|-1} \disc^{k-\tt} \rewseq_{k+1} $$

and, for the fixed trajectory $\tau$ (the subscript $\tau$ is dropped in this derivation),
$$
\rlret_{0,\tau} = \rlret_0
$$

How can these two forms be mathematically equivalent ?
- since the first form involves $\rlret_0$
    - the rewards *over all steps* of the episode
- and the second form involves $\rlret_\tt$
    - the *future* rewards from step $\tt$ onward
    - and $\rlret_\tt$ and $\rlret_{\tt'}$ for $\tt' > \tt$ include the same rewards
   
Algebraically they appear different.

The answer is that
- the two forms appear *within an expectation*
- which is evaluated over *future* time steps
- so the part of $\rlret_0$ that contains *past rewards* is equal to $0$ under the expectation

Here are the details:

- Step 1: Start with the total return $ \rlret_{0,\tau} = \rlret_0 $

Define $L$ to be the expression for the first form of the Theorem:

$
L = \sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt) \cdot \rlret_0
= \rlret_0 \sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt)
$

---

- Step 2: Expand $ \rlret_0 $ as the sum over rewards

$
\rlret_0 = \sum_{k=0}^{|\tau|-1} \disc^k \rewseq_{k+1}
$

Substitute into $ L $:

$
L = \left(\sum_{k=0}^{|\tau|-1} \disc^k \rewseq_{k+1} \right) \left(\sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt)\right)
$

---

- Step 3: Express as a double sum

$
L = \sum_{\tt=0}^{|\tau|-1} \sum_{k=0}^{|\tau|-1} \disc^k \rewseq_{k+1} \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt)
$

---

- Step 4: Separate sums over past and future rewards relative to $ \tt $

$
L = \sum_{\tt=0}^{|\tau|-1} \left( \sum_{k=0}^{\tt-1} \disc^k \rewseq_{k+1} + \sum_{k=\tt}^{|\tau|-1} \disc^k \rewseq_{k+1} \right) \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt)
$

---

- Step 5: Rewrite future rewards shifted by $ \tt $

Define $ j = k - \tt $:

$
\sum_{k=\tt}^{|\tau|-1} \disc^k \rewseq_{k+1} = \disc^\tt \sum_{j=0}^{|\tau| - 1 - \tt} \disc^j \rewseq_{\tt+j+1} = \disc^\tt \rlret_\tt
$

---

- Step 6: Substitute back into $ L $

$
L = \sum_{\tt=0}^{|\tau|-1} \left( \sum_{k=0}^{\tt-1} \disc^k \rewseq_{k+1} + \disc^\tt \rlret_\tt \right) \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt)
$

---

**Why the past-rewards term vanishes: the score function has zero mean**

$$
\E_{\actseq \sim \pi_\theta}\bigl[ \nabla_\theta \log \pi_\theta(\actseq \mid \state) \bigr]
= \sum_{\act} \pi_\theta(\act \mid \state) \, \nabla_\theta \log \pi_\theta(\act \mid \state)
= \sum_{\act} \nabla_\theta \pi_\theta(\act \mid \state)
= \nabla_\theta \sum_{\act} \pi_\theta(\act \mid \state)
= \nabla_\theta 1 = 0
$$

- rewards before step $\tt$ are already fixed when $\actseq_\tt$ is chosen, so they multiply a zero-mean term
- the same fact will justify subtracting a baseline (next session)


- Step 7: Expectation zeroes out past rewards term

Because rewards before time $\tt$ are fixed before $\actseq_\tt$ is chosen, and the score function has zero mean (previous slide), their expected contribution is zero:

$
\E\left[ \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt) \cdot \sum_{k=0}^{\tt-1} \disc^k \rewseq_{k+1} \right] = 0
$

---

- Step 8: Final form of the policy gradient

Thus,

$
\nabla_\theta J(\theta) = \E_\pi \left[ \sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt) \cdot \disc^\tt \rlret_\tt \right]
$

which is the second form of expressing the Policy Gradient Theorem.


### Alternate Proof of the Policy Gradient Theorem (Per-Step Reward Perspective)

Let $ J(\theta) $ be the expected discounted sum of rewards under policy $ \pi_\theta $:

$
J(\theta) = \E_{\tau \sim \pi_\theta} \left[ \sum_{\tt=0}^{|\tau|-1} \disc^{\tt} \rewseq_{\tt+1} \right]
$

---

- Step 1: Expand the Expectation

Rewrite the expectation explicitly:

$
J(\theta) = \sum_{\tau} \prc{\tau}{\theta} \left( \sum_{\tt=0}^{|\tau|-1} \disc^\tt \rewseq_{\tt+1} \right)
$

---

- Step 2: Differentiation w.r.t. $\theta$

$
\nabla_\theta J(\theta) = \sum_{\tau} \nabla_\theta \prc{\tau}{\theta} \left( \sum_{\tt=0}^{|\tau|-1} \disc^\tt \rewseq_{\tt+1} \right)
$

Apply the **likelihood ratio trick**:
$
\nabla_\theta \prc{\tau}{\theta} = \prc{\tau}{\theta} \nabla_\theta \log \prc{\tau}{\theta}
$
So,
$
\nabla_\theta J(\theta) = \sum_{\tau} \prc{\tau}{\theta} \nabla_\theta \log \prc{\tau}{\theta} \left( \sum_{\tt=0}^{|\tau|-1} \disc^\tt \rewseq_{\tt+1} \right)
$
Or equivalently,
$
\nabla_\theta J(\theta) = \E_{\tau \sim \pi_\theta} \left[ \nabla_\theta \log \prc{\tau}{\theta} \left( \sum_{\tt=0}^{|\tau|-1} \disc^\tt \rewseq_{\tt+1} \right) \right]
$

---

- Step 3: Break Down $\log \prc{\tau}{\theta}$

Recall,
$
\log \prc{\tau}{\theta} = \sum_{\tt=0}^{|\tau|-1} \log \pi_\theta(\actseq_\tt \vert \stateseq_\tt) + \text{terms independent of } \theta
$
So,
$
\nabla_\theta \log \prc{\tau}{\theta} = \sum_{\tt'=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_{\tt'}|\stateseq_{\tt'})
$
Substitute:

$
\nabla_\theta J(\theta) = \E_{\tau \sim \pi_\theta} \left[ \sum_{\tt'=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_{\tt'}|\stateseq_{\tt'}) \cdot \left( \sum_{\tt=0}^{|\tau|-1} \disc^\tt \rewseq_{\tt+1} \right) \right]
$

---

- Step 4: Swap Order of Summation

$
\nabla_\theta J(\theta) = \E_{\tau \sim \pi_\theta} \left[ \sum_{\tt'=0}^{|\tau|-1} \sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_{\tt'}|\stateseq_{\tt'}) \disc^\tt \rewseq_{\tt+1} \right]
$

Switch the order:

$
= \E_{\tau \sim \pi_\theta} \left[ \sum_{\tt=0}^{|\tau|-1} \sum_{\tt'=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_{\tt'} | \stateseq_{\tt'}) \disc^\tt \rewseq_{\tt+1} \right]
$

---

-  Step 5: Analyze the causal relationship

The gradient w.r.t. $\actseq_{\tt'}$ can only affect rewards *from* $\tt'$ onward (not earlier rewards: causality), so for $\tt < \tt'$ the expectation is zero.

Thus, the only contributing terms are where $\tt' \leq \tt$:

$
\nabla_\theta J(\theta) = \E_{\tau \sim \pi_\theta} \left[ \sum_{\tt'=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_{\tt'} | \stateseq_{\tt'}) \left( \sum_{\tt=\tt'}^{|\tau|-1} \disc^\tt \rewseq_{\tt+1} \right) \right]
$

---

- Step 6: Recognize the return-to-go term

$
\sum_{\tt=\tt'}^{|\tau|-1} \disc^\tt \rewseq_{\tt+1} = \disc^{\tt'} \sum_{j=0}^{|\tau|-1-\tt'} \disc^j \rewseq_{\tt'+j+1} = \disc^{\tt'} \rlret_{\tt'}
$

So, we may write:

$
\nabla_\theta J(\theta) = \E_{\tau \sim \pi_\theta} \left[ \sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt) \cdot \disc^\tt \rlret_\tt \right]
$
Many references and implementations drop $\disc^\tt$ (it is $1$ when $\disc = 1$); the exact gradient of the discounted objective keeps it.

---

- Step 7: Final form (policy gradient theorem)

$
\nabla_\theta J(\theta) = \E_{\tau \sim \pi_\theta} \left[ \sum_{\tt=0}^{|\tau|-1} \nabla_\theta \log \pi_\theta(\actseq_\tt | \stateseq_\tt) \, \disc^\tt \rlret_\tt \right]
$

---

**Conclusion:**  
By starting with the expectation of per-step rewards and applying the likelihood ratio trick, we arrive at the same policy gradient theorem:  
each action's gradient is weighted by the return-to-go from that time (not just the immediate reward).





# Actor-Critic: Policy/Value hybrid methods

**Why this slide?** A preview of where the next sessions go: policy methods can also use a value as a helper.

In classical Value-based methods

There are Policy-based methods that *also* make use of a Value function
- as a *feedback* mechanism to improve the Agent's  policy updates 

These are called Actor-Critic methods
- Actor: the Agent/Learner, implemented as a NN
- Critic: a separate NN that learns a Value function
    - "critiques" (provides feedback) on the Agent's actions

# What we established, and what it sets up

- **Established**
    - policy-based methods learn $\pi_\theta$ directly: needed for continuous actions, preferred for huge action or state spaces
    - $J(\theta)$ is the expected return; its gradient is an expectation over episodes of $\nabla_\theta \log \pi_\theta$ times a return
    - policy gradients are *on-policy*: the episodes come from the policy being improved
    - the log of an episode's probability is a sum of logs, and the transition terms do not depend on $\theta$, so they drop out of the gradient (the likelihood-ratio trick turns the gradient into an expectation): it can be estimated from sampled episodes
    - the two forms of the theorem agree in expectation (with the factor $\disc^\tt$ in the return-to-go form)
- **Sets up**
    - next session: the raw return is a noisy weight; baselines and advantages reduce the noise
    - then the practical algorithms: REINFORCE, PPO and GRPO
