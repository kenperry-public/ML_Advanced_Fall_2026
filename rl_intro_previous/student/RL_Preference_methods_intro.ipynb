# Preferences vs Rewards

There are problems where providing *exact scalar rewards*
- is harder than *ranking* potential outputs.

For example
- I may prefer chocolate to vanilla
- but I can't quantify how much more

A reward is a number: any two outputs can be compared, and the size of the gap means something.

A preference is a comparison
- "is output A better than output B?"
- we usually observe only *some* pairs, and no magnitude

        Good  > Bad
        Big   > Small
        Good vs Small ?   not comparable


Problems related to aligning the *style* of an LLM's output
is a case of preferences.
- multiple answers may be "correct"
- but one answer may be "preferred"

For example

**Prompt:** "How do I change a tire?"
- **Reply A:** An accurate step-by-step answer.
- **Reply B:** A brief, incomplete answer.

Both replies are "correct" but the first is subjectively better.

An example of *Preference Data* is a triple

$$
(x, y^+, y^-)
$$
- input $x$
- the preferred output $y^+$
- the non-preferred output $y^-$


Although both $y^+$ and $y^-$ may be *acceptable* responses, $y^+$ is preferred.

It is the *qualitative* nature of preferences (rather than correctness)
- that makes it natural to learn a reward from preferences and optimize it with Reinforcement Learning
- rather than imitate a single answer with Supervised Fine Tuning

This session only introduces preference data; using it (reward models, RLHF, DPO) is the subject of the Advanced course.

**From comparisons to a reward (preview)**

- a common model of a comparison, the Bradley-Terry model
    - $P(\y^+ \succ \y^-) = \sigma\bigl(\rew(\x, \y^+) - \rew(\x, \y^-)\bigr)$, with $\sigma$ the sigmoid
- fit $\rew$ so that preferred responses score higher: the learned reward model of the previous notebook
- only differences matter, which is why comparisons need no common scale


## The case for preferences

**Why this slide?** When is a comparison a better signal than a score?

| Scenario                                   | Why Preference Data?                     | Typical Example            |
|:---------------------------------------------|:------------------------------------------|:----------------------------|
| RLHF & LLM alignment                       | Human feedback easier as comparisons     | Choosing better LLM output |
| Hard-to-define or subjective “success”      | Comparisons are often easier to judge than scores | Dialogue, safety, style    |
| Biased or noisy scalar rewards              | A comparison does not depend on a score's scale  | Creative tasks, open-ended |
| Interpretability needs                      | Comparisons can come with a critique           | Transparent value alignment|
| DPO-style methods                          | Learn directly from comparisons (Advanced course) | Pairwise/choice based loss |


## Example of a Preference Dataset 

**Why this section?** A concrete example of preference data, to make the idea real.

Here is a [link](https://huggingface.co/datasets/HuggingFaceH4/ultrafeedback_binarized)
to the UltraFeedback dataset.

It is used to train an Assistant to be
- Helpful
    - answers the user's prompt; doesn't evade or decline
- Honest
    - gives a truthful answer



The methodology for constructing it is given [here](https://huggingface.co/datasets/HuggingFaceH4/ultrafeedback_binarized#dataset-card-for-ultrafeedback-binarized); open the link to see the columns (prompt, chosen, rejected).

- The authors gather prompts from many sources.

- Responses to each prompt are generated by several different language models.

- A strong AI model (GPT-4) acts as the annotator
    - it critiques each response on several aspects, e.g., helpfulness and honesty
    - it gives a numerical score

- Which creates a ranking of the responses to a prompt

**Synthetic Data** in practice


The "binarized" version of the dataset
- selects the highest-scored response as "Chosen"
- picks one of the remaining responses at random as "Rejected"

Note the use of AI feedback rather than Human Feedback.


# What we established, and what it sets up

- **Established**
    - a language model is a policy: the state is the prompt and the response so far, the action is the next token, a response is an episode
    - the reward is usually sparse (at the end) and can come from a program, a learned reward model or a language-model judge, each a proxy that can be hacked
    - RL amplifies behavior the model already produces, so supervised fine-tuning usually comes first
    - preference data records comparisons ($x$, preferred $y^+$, non-preferred $y^-$), which are often easier to collect than scores, and from which a reward model can be learned
- **Sets up (the Advanced course)**
    - learning a reward model from preferences and optimizing it (RLHF)
    - GRPO with verifiable rewards in practice
    - direct preference optimization, which learns from the comparisons without a separate reward model
