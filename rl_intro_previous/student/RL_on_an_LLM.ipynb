# LLM Next Token prediction task and Reinforcement Learning

**Why this session?** The methods so far came from games and robots. The same machinery trains language models. How does a language model become a policy?

One motivation for Preference Methods (introduced in the second half of this session)
- is to post-train an LLM
- to exhibit desirable characteristics
- which is often expressed with Preference Data
    - Preferred output vs. Non-Preferred output
    

We translate the LLM "Next Token Prediction" task to an instance of Reinforcement Learning
- in order to be able to use Reinforcement Learning for post-training an LLM

The Language Modeling task, formally, is
- Predict the Next Token conditional on the previously generated tokens of the response
- producing output $\y$ given input $\x$
    

For each output sequence $\y$ of tokens 
- We can associate a *state* corresponding to *each prefix* of $y$
    - the prompt and the prefix: $\stateseq_\tt = (\x, \y_{[0:\tt-1]})$
- An action is
    - choosing a token from the Vocabulary as the next output
    
- The policy $\pi_\theta( \act | \state)$ is thus equivalent to
$$\pi_\theta(\y_\tp | \x, \y_{[0:\tt-1]})$$

- the probability distribution of the next token, conditional on the prefix

Each sequence $\y$ becomes an episode/trajectory, with $|\y|$ steps (the episode length $|\tau|$ of the earlier notebooks).

**A response as an episode**

<img src="images/llm_episode.png" width=95%>

- the *state* is the prompt and the tokens generated so far; the *action* is the next token
- the whole response is one episode; the reward usually arrives only at the end (sparse), as on Frozen Lake


We can write the probability $\pi_\theta( \y | \x ) $
- of the *entire sequence* $\y$
- as the chained probabilities of each action given a state

$$
\pi_\theta( \y | \x ) = \prod_{\tt=1}^{|\y|} { \pi( \y_\tp | \x, \y_{[0:\tt-1]} ) }
$$

This is convenient for preference methods such as DPO (Advanced course)
- they compare the probability of a Preferred response $y^+$ with that of a Non-Preferred response $y^-$
- a whole-response probability is one number, instead of one factor per token

We do not use it further in this session.

**Where does the reward come from?** Classify by who produces the label

| Label produced by | Example | Strength | Weakness |
|:--|:--|:--|:--|
| a **program** (verifier) | the final number of a math problem; unit tests for code | exact, cheap, no human labels | only for checkable tasks |
| a **learned model** (reward model) | trained on human comparisons of chat answers | covers open-ended tasks | an imperfect proxy |
| a **language model as judge** | scores or ranks responses against a rubric | flexible, no training | has its own biases (e.g., favors long answers) |

- the reward is only a proxy for what we want, so the policy can learn to exploit it: **reward hacking** (the tour's Microduck found a strategy of this kind: unexpected, though nobody forbade it)
    - e.g., a model trained against a reward model may learn to write long, confident answers that score well but are not better
- a reward model is itself learned from preference data (the second half of this session)
- using these rewards in practice is the subject of the Advanced course


**One training step on a language model (the loop of sessions 5 to 7)**

- **Sample**: for a prompt $\x$, draw a group of $G$ responses $\y^{(1)}, \ldots, \y^{(G)}$ from the current policy $\pi_\theta$
- **Score**: each response gets one reward, $\rew(\x, \y)$, from a verifier, a reward model or a judge
- **Credit**: turn the rewards into advantages
    - PPO: a learned critic and GAE
    - GRPO: each reward minus the group mean, divided by the group standard deviation
    - one number per response, shared by all of its tokens
- **Update**: a policy-gradient step that raises the probability of tokens in high-advantage responses
    - plus a **KL penalty** that keeps $\pi_\theta$ close to the reference model $\pi_{\text{ref}}$ (the model before RL), so it cannot drift far
- repeat with fresh samples; the details are the subject of the Advanced course


**Why SFT usually comes before RL: RL amplifies what the model already does**

- RL raises the probability of responses the model **already produces some of the time** and that earn reward
- if no sampled response earns reward, there is no signal
    - in GRPO, a group of responses with identical rewards has every advantage equal to $0$ (the numerator is zero)
- so a model that essentially never produces the required format will not find it by RL
    - supervised fine-tuning (SFT: train on example responses) first puts that behavior within reach
- this is a condition, not a law
    - a strong base model may already succeed often enough to start RL directly (some recent reasoning models did)
    - and RL can still make rare combinations of known behaviors common
- contrast with a robot: Microduck starts from a random network, but its reward is dense (partial credit for staying upright), so there is always some signal


A check before spending compute (for your own review)

- sample $G$ responses for a few prompts; at least one should earn reward
- if the chance of success per response is a small $p$, most groups are all-fail, and a hit arrives about once per $1/(G p)$ groups
- once hits appear they are reinforced and the success rate takes off; the check is about cost, not a law
